# MURL v0.8 one-time final-test reproducibility

This companion notebook reproduces the final-test tables and figures **from archived outputs only**. It does not load the model, tokenize held-out examples, run generation, or invoke the evaluator. The one-time generation and scoring invariants are asserted from the immutable archive manifest.

Run from the repository root after placing the three ignored run artifacts in `runs/final/` as listed in the final-test report.

```python
# Archive-only regeneration; this helper contains no model or evaluator call.
import subprocess, sys
subprocess.run([sys.executable, 'scripts/v08_final_test_artifacts.py'], check=True)
```

```python
import csv
from pathlib import Path
report_dir = Path('reports/v0.8-final-test')
with (report_dir / 'summary.csv').open() as handle:
    summary = list(csv.DictReader(handle))
for row in summary:
    print(f"{row['split']}: validity={row['valid']}/{row['examples']}, correctness={row['correct']}/{row['valid']} valid, taxonomy macro-F1={float(row['taxonomy_macro_f1']):.6f}, first-error exact={round(float(row['first_error_exact']) * int(row['first_error_support']))}/{row['first_error_support']}")
```

```python
tables = sorted(path.name for path in report_dir.glob('*.csv'))
figures = sorted(path.name for path in (report_dir / 'figures').glob('*.png'))
print('Generated tables:', ', '.join(['artifact-manifest.json'] + tables))
print('Generated figures:', ', '.join(figures))
```
